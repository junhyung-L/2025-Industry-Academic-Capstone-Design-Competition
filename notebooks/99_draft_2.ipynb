지역별 결제금액 추이(전체금액/인구수)

```python
# ✅ 1. 필수 라이브러리
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ✅ 2. 파일 불러오기
df = pd.read_csv("인천e음_통합_월별.csv", encoding='utf-8')

# ✅ 3. '일자' 문자열 → datetime으로 변환
df["일자"] = pd.to_datetime(df["일자"], format="%b-%y")

# ✅ 4. 남성/여성 인구 컬럼 목록 추출
male_cols = [col for col in df.columns if col.startswith("남/총인구수")]
female_cols = [col for col in df.columns if col.startswith("여/총인구수")]

# ✅ 5. 전체 인구수 계산 및 결측 제거
df["전체인구수"] = df[male_cols + female_cols].sum(axis=1)
df = df[df["전체인구수"] > 0]  # 인구수가 0이 아닌 행만 사용

# ✅ 6. 인당 결제금액 계산
df["인당_결제금액"] = df["전체금액"] / df["전체인구수"]

# ✅ 7. 분석 기간 필터링 (2020년 7월 ~ 2025년 2월)
df_filtered = df[(df["일자"] >= "2020-07-01") & (df["일자"] <= "2025-02-28")]

# ✅ 8. 지역별-월별 평균 인당 결제금액 계산
region_monthly = df_filtered.groupby(["일자", "지역"])["인당_결제금액"].mean().reset_index()

region_monthly.to_csv("지역별_평균인당결제금액.csv", index=False, encoding='utf-8-sig')

# ✅ 9. 시각화
plt.figure(figsize=(14, 7))
for region in region_monthly["지역"].unique():
    subset = region_monthly[region_monthly["지역"] == region]
    plt.plot(subset["일자"], subset["인당_결제금액"], label=region)

plt.title("지역별 인당 결제금액 추이 (2020.07 ~ 2025.02)")
plt.xlabel("월")
plt.ylabel("인당 결제금액")
plt.legend(loc='upper right', bbox_to_anchor=(1.15, 1))
plt.tight_layout()
plt.show()
```

```python
# 기본 패키지
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
import umap.umap_ as umap

# 파일 불러오기
df = pd.read_csv("인천e음_통합_일별.csv", encoding='utf-8')
df['일자'] = pd.to_datetime(df['일자'])

# Hurst 지수 계산 함수
def hurst(ts):
    N = len(ts)
    if N < 2:
        return np.nan
    Y = np.cumsum(ts - np.mean(ts))
    R = np.max(Y) - np.min(Y)
    S = np.std(ts)
    if S == 0:
        return np.nan
    return np.log(R / S) / np.log(N)

# 지역별 Hurst 지수 계산
grouped = df[['일자', '지역', '전체금액']].dropna()
hurst_by_region = grouped.groupby('지역')['전체금액'].apply(hurst).reset_index()
hurst_by_region.columns = ['지역', 'Hurst']

# 중복 데이터 제거 후 피벗
grouped_unique = grouped.groupby(['일자', '지역'], as_index=False)['전체금액'].sum()
pivot_df = grouped_unique.pivot(index='일자', columns='지역', values='전체금액').fillna(0).T

# 정규화 + UMAP 적용
scaler = StandardScaler()
pivot_scaled = scaler.fit_transform(pivot_df)

reducer = umap.UMAP(random_state=42)
embedding = reducer.fit_transform(pivot_scaled)

# 결과 데이터프레임 구성
umap_df = pd.DataFrame(embedding, columns=['UMAP1', 'UMAP2'])
umap_df['지역'] = pivot_df.index
umap_df = umap_df.merge(hurst_by_region, on='지역', how='left')

# UMAP 시각화
plt.figure(figsize=(10, 6))
scatter = plt.scatter(
    umap_df['UMAP1'], umap_df['UMAP2'],
    c=umap_df['Hurst'], cmap='coolwarm', s=80
)
plt.colorbar(scatter, label='Hurst 지수')
plt.title("지역별 UMAP 시각화 (Hurst 지수 기반)")
plt.xlabel("UMAP1")
plt.ylabel("UMAP2")
plt.grid(True)
plt.show()

# Hurst 지수 출력 (정렬)
hurst_sorted = hurst_by_region.sort_values('Hurst', ascending=False)
print(hurst_sorted)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
import umap.umap_ as umap
from sklearn.cluster import KMeans

# ✅ 데이터 불러오기
df = pd.read_csv("인천e음_통합_일별.csv", encoding='utf-8')
df['일자'] = pd.to_datetime(df['일자'])

# ✅ Hurst 지수 계산 함수
def hurst(ts):
    N = len(ts)
    if N < 2:
        return np.nan
    Y = np.cumsum(ts - np.mean(ts))
    R = np.max(Y) - np.min(Y)
    S = np.std(ts)
    if S == 0:
        return np.nan
    return np.log(R / S) / np.log(N)

# ✅ 지역별 Hurst 지수 계산
grouped = df[['일자', '지역', '전체금액']].dropna()
hurst_by_region = grouped.groupby('지역')['전체금액'].apply(hurst).reset_index()
hurst_by_region.columns = ['지역', 'Hurst']

# ✅ 중복 제거 후 피벗 테이블
grouped_unique = grouped.groupby(['일자', '지역'], as_index=False)['전체금액'].sum()
pivot_df = grouped_unique.pivot(index='일자', columns='지역', values='전체금액').fillna(0).T

# ✅ 정규화 및 UMAP 적용
scaler = StandardScaler()
pivot_scaled = scaler.fit_transform(pivot_df)

reducer = umap.UMAP(random_state=42)
embedding = reducer.fit_transform(pivot_scaled)

# ✅ KMeans 클러스터링 (3개 군집 가정)
kmeans = KMeans(n_clusters=3, random_state=42)
cluster_labels = kmeans.fit_predict(embedding)

# ✅ 결과 데이터프레임 구성
umap_df = pd.DataFrame(embedding, columns=['UMAP1', 'UMAP2'])
umap_df['지역'] = pivot_df.index
umap_df = umap_df.merge(hurst_by_region, on='지역', how='left')
umap_df['클러스터'] = cluster_labels

# ✅ 시각화 (지역명 라벨 포함)
plt.figure(figsize=(12, 7))
scatter = plt.scatter(
    umap_df['UMAP1'], umap_df['UMAP2'],
    c=umap_df['Hurst'], cmap='coolwarm', s=100, edgecolors='k'
)
for _, row in umap_df.iterrows():
    plt.text(row['UMAP1'] + 0.1, row['UMAP2'], row['지역'], fontsize=9)
plt.colorbar(scatter, label='Hurst 지수')
plt.title("지역별 UMAP 시각화 + Hurst 지수 + 클러스터")
plt.xlabel("UMAP1")
plt.ylabel("UMAP2")
plt.grid(True)
plt.tight_layout()
plt.show()

# ✅ 저장할 데이터프레임
save_df = umap_df[['지역', 'UMAP1', 'UMAP2', 'Hurst', '클러스터']]
save_df.to_csv("hurst지수.csv", index=False, encoding="utf-8-sig")

```

```python
import pandas as pd
from statsmodels.tsa.seasonal import seasonal_decompose
import os
import unicodedata, re

# 데이터 불러오기
df = pd.read_csv("인천e음_통합_일별.csv", encoding='utf-8')
df = df[["일자", "지역", "전체금액"]]
df['일자'] = pd.to_datetime(df['일자'])

# 지역명 정규화
def clean_region_name(x):
    if pd.isnull(x):
        return x
    x = str(x).strip()
    x = unicodedata.normalize("NFKC", x)
    x = re.sub(r"\s+", " ", x)
    return x

df["지역"] = df["지역"].apply(clean_region_name)
regions = sorted(df["지역"].drop_duplicates())

# 저장 폴더 생성
output_img_dir = "시계열분해_개별구"
output_csv_dir = "시계열분해_결과CSV"
os.makedirs(output_img_dir, exist_ok=True)
os.makedirs(output_csv_dir, exist_ok=True)

# 지역별 시계열 분해 + 이미지 + CSV 저장
for region in regions:
    region_df = df[df["지역"] == region].copy()

    region_df = region_df.groupby("일자", as_index=False)["전체금액"].mean()

    region_df.set_index("일자", inplace=True)
    region_df = region_df.asfreq("D")
    region_df["전체금액"] = region_df["전체금액"].interpolate()

    # 시계열 분해
    result = seasonal_decompose(region_df["전체금액"], model='additive', period=365)

    # 📊 시각화 저장
    fig, axes = plt.subplots(1, 4, figsize=(24, 4))
    fig.suptitle(f"{region} 시계열 분해 결과", fontsize=16)

    axes[0].plot(result.observed); axes[0].set_title("전체금액")
    axes[1].plot(result.trend); axes[1].set_title("Trend")
    axes[2].plot(result.seasonal); axes[2].set_title("Seasonal")
    axes[3].plot(result.resid); axes[3].set_title("Residual")

    for ax in axes:
        ax.set_xlabel("연도")
        ax.xaxis.set_major_locator(plt.MaxNLocator(5))

    plt.tight_layout()
    plt.savefig(os.path.join(output_img_dir, f"{region}_decomp.png"))
    plt.close()

    # 📄 CSV 저장
    result_df = pd.DataFrame({
        "일자": region_df.index,
        "Observed": result.observed,
        "Trend": result.trend,
        "Seasonal": result.seasonal,
        "Residual": result.resid
    }).reset_index(drop=True)

    result_df.to_csv(os.path.join(output_csv_dir, f"{region}_decomp.csv"), index=False, encoding="utf-8-sig")
```

```python
import pandas as pd
import numpy as np
import lightgbm as lgb
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error

# ✅ 데이터 로딩
df = pd.read_csv("인천e음_통합_일별.csv", parse_dates=["일자"])
df = df.sort_values(['지역', '일자'])
df["월"] = df["일자"].dt.to_period("M").astype(str)

# ✅ 캐시백 비율 수치형 컬럼 추가
df["캐시백_3억이하"] = df["연매출 3억원 이하"].str.replace("%", "").astype(float) / 100
df["캐시백_3억~30억"] = df["연매출 3억~30억원"].str.replace("%", "").astype(float) / 100
df["캐시백_조합"] = df["캐시백_3억이하"].astype(str) + "_" + df["캐시백_3억~30억"].astype(str)

# ✅ 시간 및 정렬
df = df.sort_values(["지역", "일자"])
df["월"] = df["일자"].dt.to_period("M").astype(str)

# ✅ lag 변수 생성
for lag in range(1, 30):
    df[f"lpf_trend_shift_{lag}"] = df.groupby("지역")["lpf_trend"].shift(lag)
    df[f"금액_deviation_30_shift_{lag}"] = df.groupby("지역")["금액_deviation_30"].shift(lag)

# ✅ 변수 설정
lags = [f"lpf_trend_shift_{lag}" for lag in range(1, 30)] + \
       [f"금액_deviation_30_shift_{lag}" for lag in range(1, 30)]

base_features = [
    "평년_월별_차이", "is_weekend", "is_holiday", "month_sin", "month_cos", "day_sin", "day_cos",
    "연초말_EMT_encoded", "소비자심리지수", "CD수익률", "금리수준전망CSI", "현재생활형편CSI",
    "취업기회전망CSI", "현재가계부채CSI", "남/20_29", "남/30_39", "남/40_49", "남/50_59", "남/총인구수",
    "여/20_29", "여/30_39", "여/40_49", "여/50_59", "여/총인구수", "충전건수", "충전금액", "결제건수",
    "캐시백_3억이하", "캐시백_3억~30억"
]
features = base_features + lags
target = "전체금액"

# ✅ LightGBM 하이퍼파라미터
model_params = {
    "n_estimators": 879,
    "learning_rate": 0.006593162865614713,
    "num_leaves": 215,
    "max_depth": 10,
    "min_child_samples": 92,
    "subsample": 0.7186900301226251,
    "colsample_bytree": 0.5586715762310078,
    "reg_alpha": 1.377046983677005e-07,
    "reg_lambda": 1.8694568958308537e-08,
    "random_state": 42
}

# ✅ 결과 저장
results = []
monthly_predictions = {}

# ✅ 예측할 캐시백 비율
target_cashback = ["0.1_0.05", "0.1_0.1"]
regions = df["지역"].unique()

# ✅ 각 비율조합에 대해 지역별로 모델 학습
for cb_ratio in target_cashback:
    df_cb = df[df["캐시백_조합"] == cb_ratio]
    for region in regions:
        df_train = df_cb[(df_cb["지역"] == region) & (df_cb["일자"] < "2024-01-01")]
        df_val = df[(df["지역"] == region) & (df["일자"] >= "2024-01-01")]
        if len(df_train) < 180 or df_val.empty:
            continue

        X_train, y_train = df_train[features], df_train[target]
        X_val = df_val[features]

        model = lgb.LGBMRegressor(**model_params)
        model.fit(X_train, y_train)

        y_pred = model.predict(X_val)
        df_val_result = df_val.copy()
        df_val_result["예측값"] = y_pred
        df_val_result["비율조합"] = cb_ratio

        key = (region, cb_ratio)
        monthly_predictions[key] = df_val_result

        # 성능 측정
        mae = mean_absolute_error(df_val[target], y_pred)
        rmse = np.sqrt(mean_squared_error(df_val[target], y_pred))
        valid = df_val[target] != 0
        mape = (np.abs((df_val.loc[valid, target] - y_pred[valid]) / df_val.loc[valid, target])).mean() * 100 if valid.sum() > 0 else np.nan

        results.append({
            "비율조합": cb_ratio,
            "지역": region,
            "MAE": mae,
            "RMSE": rmse,
            "MAPE (%)": mape
        })

# ✅ 통합 테이블 출력
df_results = pd.DataFrame(results)
print(df_results)

# ✅ 시각화 및 월별 집계 비교
for region in regions:
    if (region, "0.1_0.05") in monthly_predictions and (region, "0.1_0.1") in monthly_predictions:
        df_005 = monthly_predictions[(region, "0.1_0.05")].copy()
        df_010 = monthly_predictions[(region, "0.1_0.1")].copy()

        df_005["월"] = df_005["일자"].dt.to_period("M").astype(str)
        df_010["월"] = df_010["일자"].dt.to_period("M").astype(str)

        # ✅ 일별 시각화
        plt.figure(figsize=(14, 5))
        plt.plot(df_005["일자"], df_005["예측값"], label="예측 (0.1_0.05)", linestyle="-", marker='x')
        plt.plot(df_010["일자"], df_010["예측값"], label="예측 (0.1_0.1)", linestyle="-", marker='o')
        plt.plot(df_005["일자"], df_005["전체금액"], label="실제", linestyle="--")
        plt.title(f"[일별 예측] 지역: {region} / 캐시백: 0.1_0.05 vs 0.1_0.1")
        plt.xlabel("일자")
        plt.ylabel("전체금액")
        plt.legend()
        plt.grid(True)
        plt.tight_layout()
        plt.show()

        # ✅ 월별 집계 비교
        m005 = df_005.groupby("월")[["전체금액", "예측값"]].sum().rename(columns={"예측값": "예측_0.1_0.05", "전체금액": "실제"})
        m010 = df_010.groupby("월")[["예측값"]].sum().rename(columns={"예측값": "예측_0.1_0.1"})

        df_month_compare = pd.concat([m005, m010], axis=1)
        df_month_compare["차이"] = df_month_compare["예측_0.1_0.1"] - df_month_compare["예측_0.1_0.05"]
        df_month_compare = df_month_compare.reset_index()

        # ✅ 월별 시각화
        plt.figure(figsize=(14, 5))
        plt.plot(df_month_compare["월"], df_month_compare["실제"], label="실제", marker='o')
        plt.plot(df_month_compare["월"], df_month_compare["예측_0.1_0.05"], label="예측 0.1_0.05", marker='x')
        plt.plot(df_month_compare["월"], df_month_compare["예측_0.1_0.1"], label="예측 0.1_0.1", marker='s')
        plt.title(f"[월별 집계] 지역: {region} / 캐시백 비교")
        plt.xlabel("월")
        plt.ylabel("전체금액 (합산)")
        plt.legend()
        plt.grid(True)
        plt.tight_layout()
        plt.show()

        # ✅ 차이 표 출력
        display_df = df_month_compare[["월", "예측_0.1_0.1", "예측_0.1_0.05", "차이"]]
        print(f"\n📊 지역: {region} - 월별 예측 비교표")
        print(display_df)
```

```python
import pandas as pd
from itertools import combinations

# ✅ [1] 지역/조합별 예측 결과를 월별로 집계해서 저장
monthly_list = []
for (region, cb_ratio), df_val_result in monthly_predictions.items():
    df = df_val_result.copy()
    df["월"] = df["일자"].dt.to_period("M").astype(str)
    monthly = df.groupby("월")[["전체금액", "예측값"]].sum().reset_index()
    monthly["지역"] = region
    monthly["비율조합"] = cb_ratio
    monthly_list.append(monthly)

# ✅ [2] ROI 시뮬레이션 실행 (예산 잔액 가중 반영)
df_monthly = pd.read_csv("지역별_월별_예측요약.csv")
df_monthly = df_monthly[df_monthly["월"].str.startswith("2024")]

months = sorted(df_monthly["월"].unique())
month_combos = list(combinations(months, 2))
regions = df_monthly["지역"].unique()

roi_result_list = []

budget_total = 2000 * 1e8  # ✅ 예산 2000억 원
for combo in month_combos:
    roi_sum = 0
    cashback_sum = 0
    increase_sum = 0
    budget_left = budget_total

    for region in regions:
        for i, month in enumerate(months):
            row_005 = df_monthly[(df_monthly["지역"] == region) & (df_monthly["월"] == month) & (df_monthly["비율조합"] == "0.1_0.05")]
            row_010 = df_monthly[(df_monthly["지역"] == region) & (df_monthly["월"] == month) & (df_monthly["비율조합"] == "0.1_0.1")]
            if row_005.empty or row_010.empty:
                continue

            desired_rate = 0.10 if month in combo else 0.05
            pred = float(row_010["예측값"]) if month in combo else float(row_005["예측값"])
            cashback_needed = pred * desired_rate

            if budget_left >= cashback_needed:
                rate = desired_rate
                cashback = cashback_needed
                budget_left -= cashback
            else:
                rate = 0.0
                cashback = 0.0

            if i == 0:
                prev = float(row_005["전체금액"])
            else:
                prev_row = df_monthly[(df_monthly["지역"] == region) & (df_monthly["월"] == months[i - 1]) & (df_monthly["비율조합"] == "0.1_0.05")]
                prev = float(prev_row["전체금액"]) if not prev_row.empty else float(row_005["전체금액"])

            if cashback > 0:
                roi = (pred - prev) / cashback
                weight = budget_left / budget_total
                adj_roi = roi * weight
            else:
                adj_roi = 0

            roi_sum += adj_roi
            cashback_sum += cashback
            increase_sum += (pred - prev)

    roi_result_list.append({
        "조합": combo,
        "조정된 ROI": round(roi_sum, 3),
        "캐시백 총액 (억)": round(cashback_sum / 10000, 1),
        "소비 증가액 (억)": round(increase_sum / 10000, 1),
        "남은 예산 (억)": round(budget_left / 10000, 1)
    })

df_roi = pd.DataFrame(roi_result_list).sort_values(by="조정된 ROI", ascending=False)
df_roi.to_csv("상위5개조합_ROI_조정된예산반영.csv", index=False, encoding="utf-8-sig")
print("✅ 조정된 ROI (예산 잔액 가중치 포함) 저장 완료")



# ✅ [3] 상위 1개 조합의 월별 ROI 저장
best_combo = df_roi.iloc[0]["조합"]
best_months = eval(best_combo) if isinstance(best_combo, str) else best_combo

monthly_roi_list = []
for region in regions:
    for i, month in enumerate(months):
        row_005 = df_monthly[(df_monthly["지역"] == region) & (df_monthly["월"] == month) & (df_monthly["비율조합"] == "0.1_0.05")]
        row_010 = df_monthly[(df_monthly["지역"] == region) & (df_monthly["월"] == month) & (df_monthly["비율조합"] == "0.1_0.1")]
        if row_005.empty or row_010.empty:
            continue

        pred = float(row_010["예측값"]) if month in best_months else float(row_005["예측값"])
        rate = 0.10 if month in best_months else 0.05
        cashback = pred * rate

        if i == 0:
            prev = float(row_005["전체금액"])
        else:
            prev_row = df_monthly[(df_monthly["지역"] == region) & (df_monthly["월"] == months[i-1]) & (df_monthly["비율조합"] == "0.1_0.05")]
            prev = float(prev_row["전체금액"]) if not prev_row.empty else float(row_005["전체금액"])

        roi = (pred - prev) / cashback if cashback else 0
        monthly_roi_list.append({
            "지역": region,
            "월": month,
            "적용 캐시백 비율": f"{int(rate*100)}%",
            "예측 소비액": round(pred, 1),
            "전월 소비액": round(prev, 1),
            "캐시백 비용": round(cashback, 1),
            "월별 ROI": round(roi, 2)
        })

df_monthly_roi = pd.DataFrame(monthly_roi_list)
df_monthly_roi.to_csv("1위조합_월별ROI_예산반영.csv", index=False, encoding="utf-8-sig")
print("✅ 상위 1조합의 월별 ROI 저장 완료")

```

```python
# 📌 1. 필수 라이브러리 불러오기
import pandas as pd


# 📌 3. CSV 파일 불러오기
df = pd.read_csv("인천e음_통합_월별.csv", parse_dates=["일자"])

# 📌 4. 날짜 컬럼 처리: 일자 → 월 변환
df["월"] = pd.to_datetime(df["일자"], errors="coerce").dt.to_period("M").astype(str)
df = df[df["월"].notnull()]  # NaT 제거

# 📌 5. 분석 대상 설정
target_regions = ["인천광역시 연수구", "인천광역시 옹진군"]
target_categories = [
    "문화_취미_영화관", "숙박", "약국", "연료", "유통업영리_편의점", "일반휴게음식", "학원"
]

# 📌 6. 지역별 월별 업종 결제금액 정리 및 저장
for region in target_regions:
    filtered = df[df["지역"] == region]
    result = filtered.groupby("월")[target_categories].sum().reset_index()

    # 파일명에서 "인천광역시" 제거 후 저장
    filename = f"{region.replace('인천광역시 ', '')}_업종별_월별결제금액.csv"
    result.to_csv(filename, index=False)
    print(f"저장 완료: {filename}")

# 📌 7. 저장된 파일 다운로드 링크 제공
from google.colab import files
for region in target_regions:
    files.download(f"{region.replace('인천광역시 ', '')}_업종별_월별결제금액.csv")
```
